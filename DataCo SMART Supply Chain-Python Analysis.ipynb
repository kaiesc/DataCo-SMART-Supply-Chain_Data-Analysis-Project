# FEATURE ENGINEERING

## Create Time Features - Order Date

```python
df['Order Year'] = df['order date (DateOrders)'].dt.year
df['Order Month'] = df['order date (DateOrders)'].dt.month
df['Order Month Name'] = df['order date (DateOrders)'].dt.month_name()
df['Order Day'] = df['order date (DateOrders)'].dt.day
df['Order Day Name'] = df['order date (DateOrders)'].dt.day_name() 
df['Weekend/Weekday'] = np.where(df['Order Day Name'].isin(['Saturday', 'Sunday']), 'Weekend', 'Weekday')
```

## Discount Tier Bins

```python
df['discount_tier'] = pd.cut(
    df['Order Item Discount Rate'],
    bins = [-0.01, 0.0, 0.05, 0.1, 0.15, 0.2, 0.25, 1], 
    labels = [
        'No Discount',
        '1% - 5%',
        '6% - 10%',
        '11% - 15%',
        '16% - 20%',
        '21% - 25%', 
        '>25%'
    ]
)
```

## Shipping Duration

```python
df['Order Processing (days)'] = (df['shipping date (DateOrders)'] - df['order date (DateOrders)']).dt.total_seconds() / 86400
```

## Delay Gap

```python
df['Shipment Delay (days)'] = df['Days for shipping (real)'] - df['Days for shipment (scheduled)']
```

```python
"""
1. Top Departments
"""
TopDepartment = (
    df.groupby('Department Name')
        .agg(
            total_items_sold = ('Order Item Quantity', 'sum'),
            total_profit = ('Order Profit Per Order', 'sum')
        )
        .round(2)
        .reset_index()
        .sort_values(by='total_profit', ascending=False)
        .head()
)
"""
2. Top Product Categories
"""
TopProdCateg = (
    df.groupby('Category Name')
        .agg(
            total_items_sold = ('Order Item Quantity', 'sum'),
            total_profit = ('Order Profit Per Order', 'sum')
        )
        .round(2)
        .reset_index()
        .sort_values(by='total_profit', ascending=False)
        .head()
)
"""
3. Top Market
"""
MarketRank = (
    df.groupby('Market')
        .agg(
            total_items_sold = ('Order Item Quantity', 'sum'),
            total_profit = ('Order Profit Per Order', 'sum')
        )
        .round(2)
        .reset_index()
        .sort_values(by='total_profit', ascending=False)
)
"""
4. Top Order Country
"""
TopOrderCntry = (
    df.groupby('Order Country')
        .agg(
            total_items_sold = ('Order Item Quantity', 'sum'),
            total_profit = ('Order Profit Per Order', 'sum')
        )
        .round(2)
        .reset_index()
        .sort_values(by='total_profit', ascending=False)
        .head()
)
```

```python
configs = [
    (TopDepartment, 'Department Name', 'Top Departments by Profit and Margin'),
    (TopProdCateg, 'Category Name', 'Top Product Categories by Profit and Margin'),
    (MarketRank, 'Market', 'Market Rankings by Profit and Margin'),
    (TopOrderCntry, 'Order Country', 'Top Order Countries by Profit and Margin')
]

fig, axs = plt.subplots(4, 1, figsize = (12,16))
fig.suptitle('SUPPLY CHAIN PROFIT & MARGIN PERFORMANCE', fontsize=20, fontweight='bold', y=1)


def plot_combo(data, y_col, title, ax, bar_color='#9f2a63', scatter_color='#f57d15'):
    # Set Graph Title
    ax.set_title(title, fontsize = 14, fontweight = 'bold', pad = 10)
    
    # Primary Axis: Total Profit Bar Chart
    sns.barplot(
        data = data,
        x = 'total_profit',
        y = y_col,
        ax = ax,
        color = bar_color,
        edgecolor = 'black'
    )
    ax.set_xlabel('Total Profit ($)', color = bar_color, fontweight = 'bold')
    ax.set_ylabel(y_col, fontweight = 'bold')
    ax.set_xlim(0, data['total_profit'].max() * 1.0125)
    
    # Secondary Axis: Profit Margin Scatter Points
    ax_twin = ax.twiny()
    sns.scatterplot(
        data = data,
        x = 'total_items_sold',
        y = y_col,
        ax = ax_twin,
        color = scatter_color,
        edgecolor = 'black',
        s = 150,
        zorder = 5
    )
    ax_twin.set_xlabel('Number of Units Sold', color = scatter_color, fontweight = 'bold')
    ax_twin.set_xlim(0, 120000)

for i, (df_data, y_column, title_text) in enumerate(configs):
    plot_combo(
        data = df_data, 
        y_col = y_column,
        title = title_text,
        ax=axs[i]
    )

# Adjust spacing between subplots
plt.tight_layout(rect = [0, 0, 1, 1])
plt.show()
```

```python
"""
3. Track discounting strategies and price sensitivity to ensure product lines remain profitable and discount rates are optimized to match customer buying behavior.
"""
sensitivity_analysis = (
    df.groupby('discount_tier')
        .agg(
            total_orders = ('Order Id', 'nunique'),
            total_units_sold = ('Order Item Quantity', 'sum'),
            total_profit = ('Order Profit Per Order', 'sum')
    )
).round(2).reset_index()

sensitivity_analysis['avg_profit_per_unit'] = (sensitivity_analysis['total_profit'] / sensitivity_analysis['total_units_sold']).round(2)
sensitivity_analysis
```

```python
fig, ax1 = plt.subplots(figsize=(12, 8))

sns.barplot(
    data = sensitivity_analysis,
    x = 'total_profit',
    y = 'discount_tier',
    ax = ax1,
    color = '#9f2a63',
    edgecolor = 'black'
)
ax1.set_xlabel('Total Profit ($)', color = '#9f2a63', fontsize = 12, fontweight = 'bold')
ax1.set_ylabel('Discount Tier', fontsize = 12, fontweight = 'bold')

ax2 = ax1.twiny()
sns.scatterplot(
    data = sensitivity_analysis,
    x = 'avg_profit_per_unit',
    y = 'discount_tier',
    ax = ax2,
    color = '#f57d15',
    edgecolor = 'black',
    s = 200
)
ax2.set_xlabel('Average Profit per Unit ($)', color = '#f57d15', fontsize = 12, fontweight = 'bold')

plt.title('Impact of Discount Tiers on Total Profit and Unit Profitability', fontsize = 20, fontweight = 'bold', pad = 15)
plt.show()
```

### BUSINESS INSIGHTS
- Moderate discounts between **1% – 10%** generate the highest overall profit (peaking near **\\$945K** total profit) while maintaining strong unit profitability (**\\$10.94 – $11.05** per unit).
- The **No Discount tier** yields the highest average profit per unit at **\\$12.52**, but results in lower total profit (**~\\$260K**) due to reduced purchase volume.
- Higher discount rates above 10% progressively degrade per-unit returns, dropping to **\\$9.47** per unit for the **16% – 20%** tier and down to **\\$8.65** for the **21% – 25%** tier.
- Deep discounts of **21% – 25%** produce the lowest total profit among all discounted groups (**~\\$370K**), confirming that aggressive price cuts harm overall profitability without driving sufficient sales volume to offset losses.

```python
"""
2. Evaluate shipping mode efficiency and delay rates across transit options (Standard Class, First Class, Second Class, Same Day)
segmented by Customer Country to identify which shipping methods experience the highest rate of delivery schedule breaches in each region.
"""
SLAShipping = (
    df[df["Delivery Status"] != "Shipping canceled"]
    .groupby(["Customer Country", "Shipping Mode", "Delivery Status"])["Order Id"]
    .nunique()
    .unstack("Delivery Status", fill_value=0)
    .assign(
        Total_Orders=lambda x: x["Shipping on time"] + x["Advance shipping"] + x["Late delivery"],
        Shipping_Efficiency=lambda x: (
            (x["Shipping on time"] + x["Advance shipping"])/ x["Total_Orders"] * 100
        ).round(2),
        Delay_Rate=lambda x: (x["Late delivery"] / x["Total_Orders"] * 100).round(
            2
        ),
    )
    .sort_values(["Customer Country", "Delay_Rate"], ascending=[True, False])
    .reset_index()
)

SLAShipping
```

```python
CCntry = SLAShipping["Customer Country"].unique()
num_CCntry = len(CCntry)

fig, axes = plt.subplots(
    nrows = num_CCntry,
    ncols = 1,
    figsize = (12, 4 * num_CCntry),
    sharex = True,
    sharey = True
)

# Loop through each market to plot individual stacked charts
for i, cntry in enumerate(CCntry):
    ax = axes[i]

    # Filter data for current market
    ccntry_df = SLAShipping[SLAShipping["Customer Country"] == cntry].set_index("Shipping Mode")[["Advance shipping", "Shipping on time", "Late delivery"]]

    # Normalize to 100% percentages
    cntry_pct = ccntry_df.div(ccntry_df.sum(axis = 1), axis = 0) * 100

    # Plot stacked horizontal bar chart
    bars = cntry_pct.plot(
        kind = "barh",
        stacked = True,
        colormap = "viridis",
        ax = ax,
        edgecolor = "black",
        linewidth = 0.8,
        legend = (i == 0),  # Show legend only on the first subplot
    )

    # Add percentage labels inside bar segments
    for container in ax.containers:
        labels = [
            f"{v:.2f}%" if v > 0 else "" for v in container.datavalues
        ]
        ax.bar_label(
            container,
            labels = labels,
            label_type = "center",
            color = "white",
            fontsize = 12,
            fontweight = "bold",
        )

    # Subplot styling
    ax.set_title(f"Country: {cntry}", fontsize=14, fontweight="bold", pad=10)
    ax.set_ylabel("Shipping Mode", fontsize=11, fontweight="bold")
    ax.set_xlim(0, 100)

    # Format legend for the first subplot
    if i == 0:
        ax.legend(title = "Delivery Status", bbox_to_anchor = (1.02, 1), loc = "upper left", frameon = True)

# Set common X-axis label on the bottom subplot
axes[1].set_xlabel("Percentage of Total Orders (%)", fontsize = 12, fontweight = "bold")
fig.suptitle("Fulfillment Reliability & Delay Tracking by Customer Country", fontsize = 18, fontweight = "bold")
plt.tight_layout()
plt.show()
```

### BUSINESS INSIGHTS
- **First Class shipping delivers late on every single order in both EE. UU. and Puerto Rico (100% late delivery rate)**, making it the top-priority reliability issue in the entire network.
- Second Class is the next worst performer, with **~80% of orders arriving late** in both countries — and higher order volume than First Class, so it drives more total late deliveries.
- Standard Class is the most reliable mode by far, splitting roughly **40% advance shipping / 20% on-time / 40% late** — it's the **only mode with any advance-shipping orders** at all, and it also carries the **highest order volume**, making it the operational backbone of fulfillment.
- Late delivery rates are **nearly identical across EE. UU. and Puerto Rico for every shipping mode**, pointing to a carrier/mode-level issue rather than regional logistics.
